##Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

##Carregamento do Dataset

```python
df = pd.read_csv('marketing.csv')

print(df.head(10))

# Verificar informações gerais
print(df.info())
```

##Análise Exploratória

```python
#Gráfico de dispersão
plt.scatter(df['anuncios'], df['cliques'], color='pink', alpha=0.6)
plt.title('Análise Exploratória: Anúncios vs Cliques')
plt.xlabel('Quantidade de Anúncios')
plt.ylabel('Quantidade de Cliques')
plt.show()

#Analise do grafico (Thaynná): Olhando o grafico conseguimos ver que os pontos rosas tem a tendencia de subir, da esquerda para direita em uma cresecente, isso significa uma correlação positiva.
```

##Preparação dos Dados

```python
x = df[['anuncios']] #variavel idenpedente
y = df['cliques']    #o alvo o que quero prever

#Dividindo em treino e teste - treino o que eu apresento de dados e teste o que eu guardo para que ele possa analisar
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

##Treinamento do Modelo

```python
model = LinearRegression()
model.fit(x_train, y_train)

#resultados do modelo (parâmetro: y = ax + b)
print(f"Coeficiente angular: {model.coef_[0]}")
print(f"Intercepto: {model.intercept_}")
```

```python
#Visualização da reta de regressão

plt.scatter(df['anuncios'], df['cliques'], color='pink', alpha=0.4, label='Dados Reais')


#Previsão dos cliques para os valores de anúncios
previsoes = model.predict(df[['anuncios']])

plt.plot(df['anuncios'], previsoes, color='red', linewidth=2, label='Reta de Regressão')

plt.title('Modelo de Regressão Linear Treinado')
plt.xlabel('Anúncios')
plt.ylabel('Cliques')
plt.legend()
plt.show()
```
